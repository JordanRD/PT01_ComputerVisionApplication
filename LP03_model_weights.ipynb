```python
# Install (jalankan sekali saja, hapus tanda ! kalau error)
# !pip install ultralytics opencv-python pandas

from ultralytics import YOLO
import pandas as pd
import os

# ===== PATH VIDEO (SESUAIKAN) =====
video_path = "videos/Normal_Videos_606_x264.mp4"   # relatif
# atau absolute:
# video_path = r"C:\Users\LENOVO\Documents\Kuliah\computer vision\PT01_ComputerVisionApplication-\videos\Normal_Videos_606_x264.mp4"

print(f"Path: {video_path}")
print(f"Exists: {os.path.exists(video_path)}")

if not os.path.exists(video_path):
    print("File tidak ditemukan! Cek path-nya.")
    print("Isi folder videos:", os.listdir("videos") if os.path.exists("videos") else "folder videos tidak ada")
else:
    # ===== LOAD MODEL =====
    model = YOLO("yolov8n.pt")  # atau "weights/yolov8n.pt" kalau sudah ada

    # ===== DETEKSI + TRACKING =====
    print("Memproses video...")
    results = model.track(
        source=video_path,
        conf=0.4,
        persist=True,
        tracker="bytetrack.yaml",
        save=True,
        project="runs/track",
        name="hasil_deteksi",
        exist_ok=True,
        verbose=False
    )

    print("Selesai! Cek folder runs/track/hasil_deteksi/")
```
